**Ejemplo**. *La ley de Grashof permite predecir el comportamiento de rotación de un mecanismo de cuatro barras, basándose en la longitud de sus eslabones. Sea:*

* $S$, *la longitud del eslabón más corto*
* $L$, *la longitud del eslabón más largo*
* $P$ y $Q$ *las longitudes de los dos eslabones restantes*

*El criterio de Grashof establece la siguiente clasificación:*

* *Si $ S + L < P + Q $ entonces el mecanismo es de clase I y por lo menos un eslabón será capaz de realizar una revolución completa.*
* *Si $ S + L > P + Q $ el mecanismo es de clase II y ningún eslabón será capaz de completar una revolución.*
* *Si $ S + L = P + Q $ el mecanismo es de clase III y el mecanismo presentará configuraciones inciertas.*

*Desarrolla un programa que dadas las longitudes de los eslabones de un mecanismo de cuatro barras, determine y muestre si es de clase I, clase II o clase III.*

Podemos utilizar una sentencia `if-elif-else` para clasificar un mecanismo de acuerdo con la Ley de Grashof, observa el siguiente código:

```python
a = 10 # longitud de uno de los eslabones
b = 7 # longitud de otro de los eslabones
c = 4 # longitud de otro de los eslabones
d = 5 # longitud de otro de los eslabones
S = min(a,b,c,d) # longitud del eslabón más corto
L = max(a,b,c,d) # longitud del eslabón más largo
S_mas_L = S + L # Determinamos la suma de S + L
P_mas_Q = a + b + c + d - S_mas_L # la suma de P + Q es la diferencia entre 
                                  # la suma de todas las longitudes y S + L
if S_mas_L < P_mas_Q:
    print("El mecanismo es de clase I")
elif S_mas_L > P_mas_Q:
    print("El mecanismo es de clase II")
else:
    print("El mecanismo es de clase III")
```
