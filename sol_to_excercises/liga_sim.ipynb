```python
import random
```

```python
class Equipo:
    def __init__(self,nombre,valor_plantilla):
        self.valor_plantilla = valor_plantilla
        self.nombre = nombre
        self.puntos = 0
        self.goles_a_favor = 0
        self.goles_en_contra = 0
        self.partidos_ganados = 0
        self.partidos_perdidos = 0
        self.partidos_empatados = 0

class Partido:
    def __init__(self,equipo_1,equipo_2):
        self.equipo_1 = equipo_1
        self.equipo_2 = equipo_2

    def simular(self):
        valor_plantilla_1 = self.equipo_1.valor_plantilla*random.randint(4,10)
        valor_plantilla_2 = self.equipo_2.valor_plantilla*random.randint(4,10)
        if valor_plantilla_1 > valor_plantilla_2:
            self.equipo_1.puntos += 3
            goles_equipo_1 = random.randint(1,5)
            goles_equipo_2 = random.choice(range(goles_equipo_1))
            self.equipo_1.partidos_ganados += 1
            self.equipo_2.partidos_perdidos +=1 
        elif valor_plantilla_1 < valor_plantilla_2:
            self.equipo_2.puntos += 3
            goles_equipo_2 = random.randint(1,5)
            goles_equipo_1 = random.choice(range(goles_equipo_2))
            self.equipo_1.partidos_perdidos += 1
            self.equipo_2.partidos_ganados += 1
        else:
            self.equipo_1.puntos += 1
            self.equipo_2.puntos += 1
            goles_equipo_1 = goles_equipo_2 = random.randint(1,5)
            self.equipo_1.partidos_empatados += 1
            self.equipo_2.partidos_empatados += 1
            
        self.equipo_1.goles_a_favor += goles_equipo_1
        self.equipo_1.goles_en_contra += goles_equipo_2
        self.equipo_2.goles_a_favor += goles_equipo_2
        self.equipo_2.goles_en_contra += goles_equipo_1
        # print(f"Resultado: {self.equipo_1.nombre} {goles_equipo_1} - {self.equipo_2.nombre} {goles_equipo_2}")
```

```python
import itertools
```

```python
toros = Equipo("Toros", 80)
tigres = Equipo("Tigres", 100)
tiburones = Equipo("Tiburones", 60)
jaguares = Equipo("Jaguares", 75)
aguilas = Equipo("Águilas", 130)
zorros = Equipo("Zorros", 115)
leones = Equipo("Leones", 95)
partidos = itertools.permutations([toros,tigres,tiburones,jaguares,aguilas,zorros,leones], 2)

for partido in partidos:
    px = Partido(partido[0], partido[1])
    px.simular()
```

```python
for equipo in [toros,tigres,tiburones,jaguares,aguilas,zorros,leones]:
    nombre_str = f"{equipo.nombre}".center(10)
    puntos_str = f"{equipo.puntos}".center(5)
    gf_str = f"{equipo.goles_a_favor}".center(5)
    gc_str = f"{equipo.goles_en_contra}".center(5)
    dg_str = f"{equipo.goles_a_favor - equipo.goles_en_contra}".center(5)
    print(f"{nombre_str} {puntos_str} {gf_str} {gc_str} {dg_str}")
```
